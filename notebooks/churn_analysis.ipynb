### Feature Engineering

```python
# Tenure Bucket Column
```

```python
df["TenureBuckets"] = pd.cut(df["tenure"], bins = [0,12,24,36,48,60,72], labels = ["0-12 months", "13-24 months", "25-36 months", "37-48 months", "49-60 months","61-72 months"], include_lowest = True)
```

```python
# ChurnFlag numeric column
```

```python
df["ChurnFlag"] = df["Churn"].map({"Yes" : 1, "No" : 0})
```

```python
df["ChurnFlag"].value_counts()
```

```python
df["Churn"].value_counts()
```

```python
df["ChurnFlag"].dtype
```

```python
# for eight columns which could support logic equals to yes
TotalServices = (df[["PhoneService","MultipleLines", "OnlineSecurity", "OnlineBackup", "DeviceProtection", "TechSupport", "StreamingTV", "StreamingMovies"]] == "Yes").sum(axis = 1)
```

```python
TotalServices = TotalServices + (df["InternetService"] != "No").astype(int)
```

```python
df["TotalServices"] = TotalServices
```

### Exploratory Data Analysis

```python
df = pd.read_csv("../dataset/telco_customer_churn_cleaned.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
churn_by_contract = df.groupby("Contract")[["ChurnFlag", "customerID"]].agg({"ChurnFlag": "mean", "customerID": "count"})
churn_by_contract
```

```python
sns.barplot(data = churn_by_contract, x = churn_by_contract.index, y = churn_by_contract["ChurnFlag"].values)
position, _ = plt.yticks()
labels = [f"{val * 100:.0f}" for val in position]
plt.yticks(position, labels)
plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.show()
```

```python
df["TenureBuckets"].value_counts()
```

```python
churn_by_tenure = (df.groupby("TenureBuckets")["ChurnFlag"].mean())*100
churn_by_tenure
```

```python
sns.barplot(x = churn_by_tenure.index, y = churn_by_tenure.values)
plt.title("Churn Rate by Tenure Group")
plt.xlabel("Tenure Group")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation = 30, ha = "right")
plt.show()
```

```python
df["InternetService"].value_counts()
```

```python
churn_by_internet = (df.groupby("InternetService")["ChurnFlag"].mean())*100
churn_by_internet
```

```python
sns.barplot(churn_by_internet, order = churn_by_internet.sort_values(ascending = False).index)
plt.xlabel("Internet Service")
plt.ylabel("Churn Rate (%)")
plt.title("Churn Rate by Internet Service Type")
plt.show()
```

```python
df["PaymentMethod"].value_counts()
```

```python
churn_by_payment = (df.groupby("PaymentMethod")["ChurnFlag"].mean())*100
churn_by_payment
```

```python
sns.barplot(churn_by_payment, order = churn_by_payment.sort_values(ascending = False).index)
plt.title("Churn Rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation = 10, ha = "right")
plt.show()
```

```python
df["TotalServices"].value_counts()
```

```python
churn_by_services = (df.groupby("TotalServices")["ChurnFlag"].mean())*100
churn_by_services
```

```python
sns.barplot(churn_by_services)
plt.title("Churn Rate by Total Services")
plt.xlabel("Total Services")
plt.ylabel("Churn Rate (%)")
plt.show()
```

```python
Q1 = df["MonthlyCharges"].quantile(0.25)
Q3 = df["MonthlyCharges"].quantile(0.75)
print(Q1)
print(Q3)
```

```python
df["MonthlyCharges"].min()
```

```python
df["MonthlyCharges"].max()
```

```python
df["MonthlyChargesGroup"] = pd.cut(df["MonthlyCharges"] ,bins = [-float("inf"),Q1,Q3,float("inf")], labels = ["Low", "Medium", "High"])
```

```python
df["MonthlyChargesGroup"].value_counts()
```

```python
churn_by_monthly = (df.groupby("MonthlyChargesGroup")["ChurnFlag"].mean())*100
churn_by_monthly
```

```python
sns.barplot(churn_by_monthly, order = churn_by_monthly.sort_values(ascending = False).index)
plt.title("Churn Rate by Monthly Charges")
plt.xlabel("Charges Group")
plt.ylabel("Churn Rate (%)")
plt.show()
```

```python
df["SeniorCitizen"].value_counts()
```

```python
churn_by_senior = (df.groupby("SeniorCitizen")["ChurnFlag"].mean())*100
churn_by_senior
```

```python
df["Partner"].value_counts()
```

```python
churn_by_partner = (df.groupby("Partner")["ChurnFlag"].mean())*100
churn_by_partner
```

```python
df["Dependents"].value_counts()
```

```python
churn_by_dependent = (df.groupby("Dependents")["ChurnFlag"].mean())*100
churn_by_dependent
```

```python
fig, axes = plt.subplots(2,2, figsize = (12,10))

sns.barplot(churn_by_senior, ax = axes[0][0])
axes[0][0].set_title("Churn Rate by Senior Citizen")
axes[0][0].set_xlabel("Senior Citizen")
axes[0][0].set_ylabel("Churn Rate (%)")

sns.barplot(churn_by_partner, ax = axes[0][1])
axes[0][1].set_title("Churn Rate by Partner")
axes[0][1].set_xlabel("Partner")
axes[0][1].set_ylabel("Churn Rate (%)")

sns.barplot(churn_by_dependent, ax = axes[1][0])
axes[1][0].set_title("Churn Rate by Dependents")
axes[1][0].set_xlabel("Dependents")
axes[1][0].set_ylabel("Churn Rate (%)")

axes[1][1].axis("off")
fig.suptitle("Churn Rate by Demographics", fontsize = 20, fontweight = "bold")
plt.tight_layout()
plt.show()
```
